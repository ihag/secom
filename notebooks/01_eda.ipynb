```python
X_train, X_test = X.iloc[:n], X.iloc[n:]
y_train, y_test = y.iloc[:n], y.iloc[n:]

print(X_train.shape, X_test.shape)
print(y_train.shape, y_test.shape)
print(y_train['label'].value_counts())
print(y_train['label'].value_counts(normalize=True))
```

```python
# 값이 1종류뿐인 칼럼
const_cols = X_train.columns[X_train.nunique() == 1]

# 결측률 50% 초과 컬럼
missing_cols = X_train.columns[X_train.isna().mean() > 0.5]

print(len(const_cols), len(missing_cols))

drop_cols = set(const_cols) | set(missing_cols)
print(len(drop_cols))

X_train2 = X_train.drop(columns=drop_cols)
X_test2 = X_test.drop(columns=drop_cols)

median = X_train2.median()
X_train2 = X_train2.fillna(median)
X_test2 = X_test2.fillna(median)

print(X_train2.shape, X_test2.shape)
print(X_train2.isna().sum().sum(), X_test2.isna().sum().sum())
```

```python
# 데이터 드리프트

const_all = set(X.columns[X.nunique() == 1])
const_tr  = set(X_train.columns[X_train.nunique() == 1])

diff = const_tr - const_all
print(len(diff), sorted(diff))

for c in sorted(diff):
    print(c, X_train[c].unique(), X_test[c].unique())
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (confusion_matrix, classification_report,
                             recall_score, average_precision_score)
```

```python
y_tr = y_train['label']
y_te = y_test['label']

# 1. Dummy: 전부 정상이라고 찍는 모델
dummy = DummyClassifier(strategy='most_frequent')
dummy.fit(X_train2, y_tr)
pred = dummy.predict(X_test2)
print('accuracy: ', (pred == y_te).mean())
print('recall: ', recall_score(y_te, pred, pos_label=1))
```

```python
# 2. RandomForest
rf = RandomForestClassifier(n_estimators=300, class_weight='balanced',
                            random_state=42, n_jobs=1)
rf.fit(X_train2, y_tr)
pred = rf.predict(X_test2)
proba = rf.predict_proba(X_test2)[:, 1]

print(confusion_matrix(y_te, pred, labels=[-1, 1]))
print(classification_report(y_te, pred, labels=[-1, 1]))
print('PR-AUC: ', average_precision_score(y_te, proba, pos_label=1))
```

```python
import pandas as pd
cm = confusion_matrix(y_te, pred, labels=[-1, 1])
print(pd.DataFrame(cm, index=['실제 정상', '실제 불량'],
                   columns=['예측 정상', '예측 불량']))
print(y_te.value_counts())
```

```python
import numpy as np

print(pd.Series(proba).describe())
print('최대 확률: ', proba.max())

for t in [0.05, 0.1, 0.15, 0.2, 0.3]:
    p = np.where(proba >= t, 1, -1)
    n_pred = ((p == 1) & (y_te == 1)).sum()
    n_hit = ((p == 1) & (y_te == 1)).sum()
    print(f'기준 {t}: 불량으로 예측 {n_pred}개, 그중 진짜 불량 {n_hit}개')
```

```python
from sklearn.metrics import roc_auc_score

print('ROC-AUC:', roc_auc_score(y_te, proba))   # 0.5면 무작위
print(pd.Series(proba).groupby(y_te.values).mean())   # 정상(-1)/불량(1)의 평균 확률
```
